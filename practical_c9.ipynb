## Weights update formula

$`\omega_{i+1} = \omega_i + \eta\Delta\omega_i + \mu\Delta\omega_{i-1}`$

$`\Delta\omega`$ is the gradient for the error function for that weight

$`\mu`$ is the momentum (typically around 0.9)

$`\eta`$ is the learning rate, can be updated with $`\eta = \frac{\eta_0}{t^p}`$

with $`\eta_0`$ the initial value of the learning rate, usually just above 0,
like 0.0.5

t is the iteration number of batches or mini-batches

p is an exponant, typically around 0.5

## Forward pass

$`a^{(L)} = h(W^{(L)}a^{(L-1)}+b^{(L)})`$

$`a^{(L)}`$ are the activations of the nodes at layer L

$`W^{(L)}`$ is the matrix of weights between layer L-1 and L. Column i, row j
goes from node i at level L-1 to node j at level L.

$`b^{(L)}`$ are the bias at level L

$`z^{(L)} = W^{(L)}a^{(L-1)}+b^{(L)}`$ is the argument to the activation
function

h is the activation function (like a sigmoid)

### Sigmoid

If $`h(x) = \frac{1}{1+e^{-x}}`$, then $`h'(x) = h(x)(1-h(x))`$

## Backpropagation

$`\mathcal{L}`$ is the loss function $` = \frac{1}{2}(y - \^{y})^2`$

$`\delta^{(L)} = \frac{\partial \mathcal{L}}{\partial \mathcal{z}^{(L)}}`$ is
defined as the contribution to the loss from the inputs to layer L.

Then
$`\delta^{(L)} = ({W^{L+1}}^T \delta^{(L+1)}) \cdot h'(z^{(L)})`$ with the
matrix product here being multiplication of matrix elements of the same position

From there, $`\frac{\partial \mathcal{L}}{\partial b_j^{(L)}} = \delta_j^{(L)}`$
and $`\frac{\partial \mathcal{L}}{\partial w_{kj}^{(L)}} = a_k^{L-1}\delta_j^{(L)}`$
for bias $`b_j^{(L)}`$ is the bias at node j, layer L; and $`w_{kj}^{(L)}`$ is
the weight from node k at layer L-1 to node j at layer L.

## Loss functions

$`\mathcal{L}_{abs} = |y-\^{y}|`$

$`\mathcal{L}_{MSE} = \frac{1}{2}(y - \^{y})^2`$

### For classification

$`\mathcal{L}_{ent} = - \sum_{i=0}^{M}\^{y}_i log y_i`$ for M classes

$`\mathcal{L}_{ent} = -\^{y} log(y) + (1-\^{y})log(1-y)`$ in the binary case

# Weights initialization

Uniform or normal distribution within [0,1] is not good enough.

Xavier Glorot initialization is uniform within
$`[-\sqrt{\frac{A}{f_{in}+f_{out}}}, \sqrt{\frac{A}{f_{in}+f_{out}}}]`$

A is a constant, 2 for sigmoid, 6 otherwise.

$`f_{in}`$ and $`f_{out}`$ are the number of weights into and out of the layer

Alternate version is $`N(0,1)\sqrt{\frac{1}{f_{in}}}`$

N(0,1) is a normal distribution of mean 0 and std dev of 1.

For ReLU, He initialization is $`N(0,1)\sqrt{\frac{2}{f_{in}}}`$

## Regularization

To keep the learning general and not on the details of the training set.

### L2 regularization

Change the loss function to
$`\mathcal{L} = \mathcal{L}(x,y,w,b)+\frac{\lambda}{2}\sum_i w_i^2`$

### Drop out

For each sample, only take a random subset of the neural network. The weights
also need to be multiplied by that probability.
