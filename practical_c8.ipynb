## Building the Iris dataset

```python
import numpy as np

d = np.load("iris/iris_train_features_augmented.npy")
l = np.load("iris/iris_train_labels_augmented.npy")
d1 = d[np.where(l==1)]
d2 = d[np.where(l==2)]
a=len(d1)
b=len(d2)
x = np.zeros((a+b,2))
x[:a,:] = d1[:,2:]
x[a:,:] = d2[:,2:]
y = np.array([0]*a+[1]*b)
i = np.argsort(np.random.random(a+b))
x = x[i]
y = y[i]
np.save("iris/iris2_train.npy", x)
np.save("iris/iris2_train_labels.npy", y)
d = np.load("iris/iris_test_features_augmented.npy")
l = np.load("iris/iris_test_labels_augmented.npy")
d1 = d[np.where(l==1)]
d2 = d[np.where(l==2)]
a=len(d1)
b=len(d2)
x = np.zeros((a+b,2))
x[:a,:] = d1[:,2:]
x[a:,:] = d2[:,2:]
y = np.array([0]*a+[1]*b)
i = np.argsort(np.random.random(a+b))
x = x[i]
y = y[i]
np.save("iris/iris2_test.npy", x)
np.save("iris/iris2_test_labels.npy", y)
```

```python
import numpy as np
import pickle
from sklearn.neural_network import MLPClassifier

xtrain= np.load("iris/iris2_train.npy")
ytrain= np.load("iris/iris2_train_labels.npy")
xtest = np.load("iris/iris2_test.npy")
ytest = np.load("iris/iris2_test_labels.npy")

clf = MLPClassifier(
        hidden_layer_sizes=(3,2),
        activation="logistic",
        solver="adam", tol=1e-9,
        max_iter=5000,
        verbose=True)
clf.fit(xtrain, ytrain)
prob = clf.predict_proba(xtest)
score = clf.score(xtest, ytest)

w12 = clf.coefs_[0]
w23 = clf.coefs_[1]
w34 = clf.coefs_[2]
b1 = clf.intercepts_[0]
b2 = clf.intercepts_[1]
b3 = clf.intercepts_[2]
weights = [w12,b1,w23,b2,w34,b3]
pickle.dump(weights, open("iris/iris2_weights.pkl","wb"))

print()
print("Test results:")
print("  Overall score: %0.7f" % score)
print()
for i in range(len(ytest)):
    p = 0 if (prob[i,1] < 0.5) else 1
    print("%03d: %d - %d, %0.7f" % (i, ytest[i], p, prob[i,1]))
print()
```
